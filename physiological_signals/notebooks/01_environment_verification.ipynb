```python

# CELL 1 — LIBRARY VERIFICATION
import sys
print(f"Python version: {sys.version}")
print()

libraries = {}

try:
    import numpy as np
    libraries['numpy'] = np.__version__
except ImportError as e:
    libraries['numpy'] = f"FAILED: {e}"

try:
    import pandas as pd
    libraries['pandas'] = pd.__version__
except ImportError as e:
    libraries['pandas'] = f"FAILED: {e}"

try:
    import matplotlib
    libraries['matplotlib'] = matplotlib.__version__
except ImportError as e:
    libraries['matplotlib'] = f"FAILED: {e}"

try:
    import seaborn as sns
    libraries['seaborn'] = sns.__version__
except ImportError as e:
    libraries['seaborn'] = f"FAILED: {e}"

try:
    import scipy
    libraries['scipy'] = scipy.__version__
except ImportError as e:
    libraries['scipy'] = f"FAILED: {e}"

try:
    import sklearn
    libraries['scikit-learn'] = sklearn.__version__
except ImportError as e:
    libraries['scikit-learn'] = f"FAILED: {e}"

try:
    import xgboost as xgb
    libraries['xgboost'] = xgb.__version__
except ImportError as e:
    libraries['xgboost'] = f"FAILED: {e}"

try:
    import tensorflow as tf
    libraries['tensorflow'] = tf.__version__
except ImportError as e:
    libraries['tensorflow'] = f"FAILED: {e}"

try:
    import neurokit2 as nk
    libraries['neurokit2'] = nk.__version__
except ImportError as e:
    libraries['neurokit2'] = f"FAILED: {e}"

try:
    import pywt
    libraries['PyWavelets'] = pywt.__version__
except ImportError as e:
    libraries['PyWavelets'] = f"FAILED: {e}"

try:
    import shap
    libraries['shap'] = shap.__version__
except ImportError as e:
    libraries['shap'] = f"FAILED: {e}"

try:
    import biosppy
    libraries['biosppy'] = 'installed'
except ImportError as e:
    libraries['biosppy'] = f"FAILED: {e}"

try:
    import tqdm
    libraries['tqdm'] = tqdm.__version__
except ImportError as e:
    libraries['tqdm'] = f"FAILED: {e}"

try:
    import joblib
    libraries['joblib'] = joblib.__version__
except ImportError as e:
    libraries['joblib'] = f"FAILED: {e}"

print("=" * 55)
print("LIBRARY VERIFICATION RESULTS")
print("=" * 55)

all_passed = True
for lib, version in libraries.items():
    status = "PASS" if "FAILED" not in str(version) else "FAIL"
    print(f"  [{status}]  {lib}: {version}")
    if "FAILED" in str(version):
        all_passed = False

print("=" * 55)
if all_passed:
    print("ALL LIBRARIES PASSED")
    print("Environment is ready")
else:
    print("SOME FAILED — fix before proceeding")
print("=" * 55)
```
